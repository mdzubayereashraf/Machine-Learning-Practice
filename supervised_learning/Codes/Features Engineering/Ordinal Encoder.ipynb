# **Ordinal Encoding**

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
```

```python
data = [['good'],['average'], ['excilent'], ['perfect'], ['good'],
        ['average'], ['excilent'], ['perfect'],['good'],['average'], 
        ['excilent'],
        ['perfect'],['good'],['average'], ['excilent'], ['perfect']]
```

```python
df= pd.DataFrame(data =data, columns =["Reviews"])
df.head()
```

```python
df.shape
```

## Defining Order

```python
order = [['average', 'good', 'perfect', 'excilent']]
encoder = OrdinalEncoder( categories = order)

```

# Encoding Data

```python
encoded_data = encoder.fit_transform(df)
encoded_data
```

```python
decoded_data = encoder.inverse_transform(encoded_data)
decoded_data
```
