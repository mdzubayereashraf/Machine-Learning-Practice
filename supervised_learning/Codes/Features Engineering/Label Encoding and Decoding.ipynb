## **Label Encoding And Decoding**

```python
from sklearn.preprocessing import LabelEncoder
```

```python
data = ['classA','classD','classB','classD','classC','classD','classC','classD','classA','classB','classA','classD','classC','classD']
```

## Encoding

```python
label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(data)

print("Encoded Data:", encoded_labels)
```

## Decoding

```python
original = label_encoder.inverse_transform(encoded_labels)

print("Decoded Data:", original)
```
