# **One-Hot Encoding**

```python
import pandas as pd
```

```python
data ={'Categories' : ['classA','classD','classB','classD','classC','classD','classC','classD','classA','classB','classA','classD','classC','classD']}
df = pd.DataFrame(data)
df
```

```python
onehotencodedvar = pd.get_dummies(df, columns=["Categories"])
onehotencodedvar
```

```python
onehotencodedvar = pd.get_dummies(df, columns=["Categories"], prefix='Dummy',drop_first= True)
onehotencodedvar
```
