# **Normalization Process**

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    'feature1': [25, 45, 46,  23, 45, 65],
    'feature2' : [34, 43, 78, 65, 68 ,97]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
standardizer = StandardScaler()
standarddata = standardizer.fit_transform(df)
standarddata
```
